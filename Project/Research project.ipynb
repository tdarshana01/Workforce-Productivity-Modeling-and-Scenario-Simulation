#### Outlier Detection

```python
def iqr_outlier_summary_exclude(df, num_cat_cols, multiplier=1.5):
    summary = []

    # select numerical columns and exclude num_cat_cols
    numeric_cols = (
        df.select_dtypes(include='number')
          .columns
          .difference(num_cat_cols)
    )

    for col in numeric_cols:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - multiplier * iqr
        upper = q3 + multiplier * iqr

        n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()

        summary.append({
            "Variable": col,
            "Outliers_Count": int(n_outliers)
        })

    return pd.DataFrame(summary)
```

```python
outlier_table = iqr_outlier_summary_exclude(df, num_cat_cols)
outlier_table
```

#### CLuster Analysis

```python
import prince
from sklearn.cluster import KMeans
```

```python
df_famd = df.drop(columns=["PerformanceScore"])
```

```python
# Convert numerical columns to float
num_cols = df_famd.select_dtypes(include=['int64']).columns

df_famd[num_cols] = df_famd[num_cols].astype(float)
```

```python
famd = prince.FAMD(
    n_components=25,  
    random_state=42
)

famd = famd.fit(df_famd)
famd_coords = famd.transform(df_famd)
```

```python
# Eigenvalues
eig_vals = famd.eigenvalues_
print(eig_vals)
```

```python
famd.eigenvalues_summary
```

```python
famd_selected = famd_coords.iloc[:, :19]
```

```python
wcss = []

# Try different number of clusters
for i in range(2, 10):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(famd_selected)
    wcss.append(kmeans.inertia_)

# Plot
plt.plot(range(2, 10), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
#plt.title("Elbow Method for Optimal Clusters")
plt.show()
```

```python
from sklearn.metrics import silhouette_score

for i in range(2, 10):
    kmeans = KMeans(n_clusters=i, random_state=42)
    labels = kmeans.fit_predict(famd_selected)
    score = silhouette_score(famd_selected, labels)
    print(f"K={i}, Silhouette Score={round(score,2)}")
```

### Advance Analysis

```python
df1 = df.copy()
```

```python
x = df1.drop('PerformanceScore',axis=1)
y = df1['PerformanceScore']
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
import numpy as np
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
models = [LinearRegression(),SVR(),DecisionTreeRegressor(),RandomForestRegressor(),XGBRegressor(),GradientBoostingRegressor(),AdaBoostRegressor(),KNeighborsRegressor()]
```

```python
x_prep = pd.get_dummies(x)
```

```python
x_prep = x_prep.map(lambda x:1 if x == True else (0 if x == False else x))
```

```python
x_prep
```

```python
X_train_prep, X_test_prep, y_train_prep, y_test_prep = train_test_split(x_prep,y,test_size=0.2,random_state=42)
```

```python
scores = []

for model in models:
    score = cross_val_score(model,X_train_prep,y_train_prep,cv=5,n_jobs=-1)
    score = np.round(score,2)
    scores.append({
        'model':model,
        'scores':score,
        'mean_score':np.mean(score)
    })
    
pd.DataFrame(scores)
```

```python
scores = []

for model in models:
    score = cross_val_score(
        model,
        x_prep,
        y,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )
    
    rmse = -score  # Convert back to positive RMSE
    
    scores.append({
        'model': model,
        'rmse_scores': np.round(rmse, 2),
        'mean_rmse': np.round(np.mean(rmse), 2)
    })

pd.DataFrame(scores)
```
